```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
x1 = np.arange(0,10,0.1)
x2 = np.arange(0,10,0.1)

x1_dot = x1+ x1*x2
x2_dot = -x2+2*x1*x2
```

```python
X1,X2 = np.meshgrid(x1,x2)
```

```python
X1_dot = X1+ X1*X2
X2_dot = -X2+2*X1*X2
```

```python
fig = plt.figure(figsize=(12,12))
ax1 = fig.add_subplot(121, projection='3d')
ax2 = fig.add_subplot(122, projection='3d')
# Plot the surface
surf1 = ax1.plot_surface(X1, X2, X1_dot, cmap='viridis', edgecolor='none')
surf2 = ax2.plot_surface(X1, X2, X2_dot, cmap='viridis', edgecolor='none')
```

```python
contour = plt.contour(X1, X2, X1_dot, levels=20, cmap='viridis')
plt.clabel(contour, inline=True, fontsize=8)
```

```python
contour = plt.contour(X1, X2, X2_dot, levels=20, cmap='viridis')
plt.clabel(contour, inline=True, fontsize=8)
```
